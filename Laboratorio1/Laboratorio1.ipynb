```python
# 1. Dataset valores de ejemplo: Temperatura vs Consumo de electrodomésticos

x = [18.5, 19.0, 19.2, 19.5, 20.0, 20.3, 20.7, 21.0, 21.4, 21.8,
     22.0, 22.3, 22.7, 23.0, 23.2]
y = [55,   60,   58,   70,   75,   72,   85,   90,   88,   95,
     100,  105,  110,  120,  115]

n = len(x)
print(f"Total de registros: {n}")
print("Variables: x = Temperatura (°C), y = Consumo (Wh)\n")

# 2. Búsqueda de los mejores valores de b y w
# Probamos muchas combinaciones y nos quedamos con la que da menor error

mejor_error = float('inf')
mejor_b = 0
mejor_w = 0

# Rangos de búsqueda
# b: intercepto   |  w: pendiente
for b in range(-50, 151, 2):
    for w in range(-20, 31):

        error = 0.0

        # Calculamos el error total para este par (b, w)
        for i in range(n):
            y_predicho = b + w * x[i]
            diferencia = y[i] - y_predicho
            error = error + (diferencia ** 2)

        # Estructura decisiva: ¿este error es mejor que el anterior?
        if error < mejor_error:
            mejor_error = error
            mejor_b = b
            mejor_w = w

# 3. Mostrar los mejores valores encontrados
print("--- MEJORES VALORES ENCONTRADOS ---")
print(f"Intercepto (b) = {mejor_b}")
print(f"Pendiente (w)  = {mejor_w}")
print(f"Ecuación: y = {mejor_b} + {mejor_w}x")
print(f"Error mínimo (suma de cuadrados) = {mejor_error:.2f}\n")

# 4. Inferencias para 10 nuevos valores de x
nuevos_x = [18.0, 18.8, 19.5, 20.2, 20.9, 21.5, 22.1, 22.8, 23.5, 24.0]

print("--- INFERENCIAS PARA 10 NUEVOS VALORES DE X ---")
for i in range(len(nuevos_x)):
    x_nuevo = nuevos_x[i]
    y_predicho = mejor_b + mejor_w * x_nuevo
    print(f"Caso {i+1}: Si x = {x_nuevo} °C  -->  y predicho = {y_predicho:.2f} Wh")

# 5. Explicación del algoritmo y de los resultados
print("\n" + "="*60)
print("EXPLICACIÓN DEL ALGORITMO EMPLEADO")
print("="*60)
print("""
El programa NO usa fórmulas de regresión ni librerías.
En su lugar hace una búsqueda exhaustiva (prueba y error):

1. Se definen rangos posibles para b (intercepto) y w (pendiente).
2. Con dos bucles for anidados se prueba CADA combinación posible de b y w.
3. Para cada combinación se calcula el error total:
      - Se predice ŷ = b + w*x para todos los puntos
      - Se suma el cuadrado de las diferencias (y real - ŷ)
4. Con un if se guarda la combinación que produce el MENOR error.
5. Al terminar los bucles, mejor_b y mejor_w son los valores óptimos
   encontrados dentro del rango explorado.

Este método se llama "búsqueda por rejilla" (grid search) y solo
necesita estructuras de control repetitivas (for) y decisivas (if).
""")

print("="*60)
print("EXPLICACIÓN DE LOS RESULTADOS")
print("="*60)
print(f"""
Se encontró que la mejor recta dentro del rango buscado es:
    y = {mejor_b} + {mejor_w}x

Esto significa que:
- Cuando la temperatura es 0 °C, el consumo estimado sería {mejor_b} Wh
  (valor teórico del intercepto).
- Por cada grado que aumenta la temperatura, el consumo aumenta
  aproximadamente {mejor_w} Wh (pendiente).

Las 10 inferencias muestran el consumo estimado para temperaturas
que no estaban en el dataset original. Son predicciones basadas
en la recta que mejor se ajustó a los datos según el criterio
de mínimo error cuadrático.
""")
```
